```python
#program to print the factorial number
num = int(input("Enter a number:")) #7
fact = 1
for i in range(num,0,-1): #i=7,6,5,4,3,2,1
    fact = fact*i
    print(fact)
print("the factorial of given num is",fact)
```

```python
#Write a program to print given number is prime or not(basic)
num = int(input("Enter a number:"))
count = 0
for i in range(2,num,1):
    if(num%i==0):
        count = count+1
print("count=",count)
if count==0:
    print(num,"is a prime")
else:
    print(num,"is not a prime")
```

```python
#Loop Statements
break
continue
pass
```

```python
#break statement 
for i in range(1,1000,1):
    if(i==8):
        break
    else:
        print(i)
```

```python
#break statement 
for i in range(1,1000,1):
    if(i==18):
        break
    else:
        print(i)
```

```python
#continue statement 
for i in range(1,15,1):
    if(i==8):
        continue
    else:
        print(i)
```

```python
#continue statement 
for i in range(1,15,1):
    if(i==8 or i==9 or i==10):
        continue
    else:
        print(i)
```

```python
#pass statement 
for i in range(1,15,1):
    if(i==8):
        pass
    else:
        print(i)
```

```python
#Write a program to print given number is prime or not(using break statement)
num = int(input("Enter a number:"))
for i in range(2,num,1):
    if(num%i==0):
        print(num,"is a prime")
        break
else:
    print(num,"is not a prime")
```
