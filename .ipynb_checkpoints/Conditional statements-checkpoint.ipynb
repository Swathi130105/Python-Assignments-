```python
# Write a program to print result based on percentage
85 to 100 is distinction
60 to 85 is first class 
50 to 59 is second class
35 to 49 is pass
<35 is fail
```

```python
per=float(input("enter the percentage:"))
if(per>85 and per<=100):
    print("distinction")
elif(per>=60):
    print("first class")
elif(per>=50):
    print("second class")
elif(per>=35):
    print("pass")
else:
    print("fail")
```

```python
per=float(input("enter the percentage:"))
if(per>85 or per>100):
    print("invalid input")
elif(per>=85):
    print("distinction")
elif(per>=60):
    print("first class")
elif(per>=50):
    print("second class")
elif(per>=35):
    print("pass")
else:
    print("fail")
```

```python
# write a program to print its a equilater triangle,isoceles triangle, scalene triangle or not a triangle
a = float(input("enter a 1st side of triangle:"))
b = float(input("enter a 2st side of triangle:"))
c = float(input("enter a 3st side of triangle:"))
if(a+b<c or a+c<b or b+c<a):
    print("not a triangle")
elif(a==b==c):
    print("equilateral triangle")
elif(a==b or b==c or c==a):
    print("isocelese triangle")
else:
    print("scalene triangle")
    
```

```python
#Write a program for zero ,+ve ,-ve
num = float(input("enter a number:"))
if(num!=0):
    if(num>0):
        print(num,"is +ve")
    else:
        print(num,"is -ve")
else:
    print(num,"is zero")
```

```python
#Discount calculator takes the total bill amount  as input
if bill >= 5000
if bill >=10000 (gives 20% discount)
else(give 10% discount)
else
if bill>=2000(give 5% discount)
else("No discount")
```

```python
bill_amount=float(input("enter the total bill amount:")) 
if(bill_amount>=2000):
    if(bill_amount>=5000):
        if(bill_amount>=10000):
            print("20% discount")
            discount_amount=bill_amount*0.2
        else:
            print("10% discount")
            discount_amount=bill_amount*0.1
    else:
        print("5% discount")
        discount_amount=bill_amount*0.5
else:
    print("no discount")
    discount_amount=0
    
print("the discount amount is:",discount_amount)
print("the bill amount to pay is:",bill_amount-discount_amount)
```
