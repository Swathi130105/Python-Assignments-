```python
1) 1. E-commerce Discount Calculator

Scenario: An online store offers discounts based on the purchase amount:

10% discount for purchases between $100 and $500

20% discount for purchases above $500

No discount for purchases below $100
```

```python
amount = float(input("Enter purchase amount: $"))
if amount > 500:
    discount = amount * 20/100
elif amount >= 100:
    discount = amount * 10/100
else:
    discount = 0
final_amount = amount - discount
print("Purchase Amount: $", amount)
print("Discount Amount: $", discount)
print("Final Amount to Pay: $", final_amount)
```

```python
6) Electricity Bill Calculator

Scenario: An electricity company charges its customers as follows:

First 100 units: $0.5 per unit

Next 100 units (101-200): $0.75 per unit

Above 200 units: $1 per unit
```

```python
units = int(input("Enter electricity units consumed: "))
if units <= 100:
    bill = units * 0.5
elif units <= 200:
    bill = (100 * 0.5) + ((units - 100) * 0.75)
else:
    bill = (100 * 0.5) + (100 * 0.75) + ((units - 200) * 1)
print("Total Electricity Bill: $", bill)
```

1) Leap Year Checker
Accept a year. First check whether it is divisible by 4. If yes, check whether it is divisible by 100. If it is, check whether it is divisible by 400. Display whether the year is a leap year.

```python
year = int(input("Enter a year: "))
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("It is a leap year")
        else:
            print("It is not a leap year")
    else:
        print("It is a leap year")
else:
    print("It is not a leap year")
```

2) Triangle Classification
Accept three side lengths. First check whether they can form a valid triangle. If valid, check whether all three sides are equal, exactly two sides are equal, or all sides are different.

```python
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))
if a + b > c and a + c > b and b + c > a:
    if a == b:
        if b == c:
            print("Equilateral Triangle")
        else:
            print("Isosceles Triangle")
    else:
        if b == c or a == c:
            print("Isosceles Triangle")
        else:
            print("Scalene Triangle")
else:
    print("Invalid Triangle")
```

3) Employee Bonus Calculator
Accept years of experience and performance rating. If experience is at least 5 years, check the rating. Give a 20% bonus for an excellent rating and 10% for a good rating. If experience is below 5 years, give a 5% bonus only for an excellent rating.

4) Password Strength Checker
Accept a password. If its length is at least 8 characters, check whether it contains a digit. If it does, check whether it contains an uppercase letter. Classify it as strong or moderate based on these conditions. Otherwise, classify it as weak.

```python
password = input("Enter your password: ")
if len(password) >= 8:
    if any(char.isdigit() for char in password):
        if any(char.isupper() for char in password):
            print("Strong Password")
        else:
            print("Moderate Password")
    else:
        print("Weak Password")
else:
    print("Weak Password")
```

6) Income Tax Slab Calculator
Accept annual taxable income. If income is up to ₹3 lakh, tax is 0%. Otherwise, check whether it is up to ₹7 lakh and apply a 5% rate to income above ₹3 lakh. For income above ₹7 lakh, apply progressive slabs of 0%, 5%, and 10% to the respective portions. Calculate the total tax.

```python
income = float(input("Enter annual taxable income: ₹"))
if income <= 300000:
    tax = 0
else:
    if income <= 700000:
        tax = (income - 300000) * 0.05
    else:
        tax = (400000 * 0.05) + ((income - 700000) * 0.10)
print("Annual Taxable Income: ₹", income)
print("Total Income Tax: ₹", tax)
```

7) College Admission
Accept marks in Mathematics, Physics, and Chemistry. If Mathematics is at least 60, check Physics. If Physics is at least 50, check Chemistry. If Chemistry is at least 50 and the total is at least 180, display "Eligible"; otherwise, display "Not Eligible".

```python
maths = int(input("Enter Mathematics marks: "))
physics = int(input("Enter Physics marks: "))
chemistry = int(input("Enter Chemistry marks: "))
total = maths + physics + chemistry
if maths >= 60:
    if physics >= 50:
        if chemistry >= 50:
            if total >= 180:
                print("Eligible")
            else:
                print("Not Eligible")
        else:
            print("Not Eligible")
    else:
        print("Not Eligible")
else:
    print("Not Eligible")
```

8)Online Food Delivery Accept order amount, distance, and whether the customer has a membership. If the order amount is at least ₹500, check whether the distance is within 5 km. Apply free delivery if within 5 km; otherwise, charge ₹40. If the order is below ₹500, check membership status and calculate delivery charges accordingly.

```python
amount = float(input("Enter order amount: ₹"))
distance = float(input("Enter delivery distance in km: "))
membership = input("Do you have membership? (yes/no): ").lower()
if amount >= 500:
    if distance <= 5:
        delivery_charge = 0
    else:
        delivery_charge = 40
else:
    if membership == "yes":
        delivery_charge = 20
    else:
        delivery_charge = 50

total = amount + delivery_charge
print("Order Amount: ₹", amount)
print("Delivery Charge: ₹", delivery_charge)
print("Total Amount: ₹", total)
```

9)Insurance Premium Category
Accept age, smoking status, and annual income. If the applicant is under 30, check smoking status to determine the risk category. If age is 30 or above, check income and smoking status to classify the applicant as low, medium, or high risk. Define your own clear rules for each category.

```python
account = input("Enter account type (savings/current): ").lower()
balance = float(input("Enter account balance: "))
transaction = input("Enter transaction (deposit/withdrawal): ").lower()
amount = float(input("Enter transaction amount: "))

if account == "savings":
    if transaction == "deposit":
        if amount > 0:
            balance = balance + amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")
    elif transaction == "withdrawal":
        if balance - amount >= 1000 and amount > 0:
            balance = balance - amount
            print("Withdrawal successful")
        else:
            print("Minimum balance requirement not met or invalid amount")
    else:
        print("Invalid transaction type")

elif account == "current":
    if transaction == "deposit":
        if amount > 0:
            balance = balance + amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")
    elif transaction == "withdrawal":
        if balance - amount >= 5000 and amount > 0:
            balance = balance - amount
            print("Withdrawal successful")
        else:
            print("Minimum balance requirement not met or invalid amount")
    else:
        print("Invalid transaction type")
else:
    print("Invalid account type")

print("Current balance: ₹", balance)
```
